```python
import pyarrow.parquet as pq
import pandas as pd
```

```python
#read in data
files = ['yellow_tripdata_2025-06.parquet', 'yellow_tripdata_2025-07.parquet']

tables = [pq.read_table(f).to_pandas() for f in files]
df = pd.concat(tables, ignore_index=True)
print(df.info())
```

```python
print("Head:")
df.head(10)
```

```python
print("Missing Values:")
df.isna().sum()
```

```python
print("dropoff before pickup:", len(df[df["tpep_dropoff_datetime"] < df["tpep_pickup_datetime"]]))
print("negative fare amount length:", len(df[df["fare_amount"]<0]))
print("flex fare (1), no charge (3), disputed (4), unknown (5), and voided charges (6)")
df[(df["payment_type"]!=1) & (df["payment_type"]!=2)].head(10)
```

```python
print("Difference in amounts <= 100 and greater than 100")
len(df[df["total_amount"]<=100]) - len(df[df["total_amount"]>100])
```
